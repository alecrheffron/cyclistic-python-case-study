```python
for threshold in [60, 120, 180, 360, 720, 1440]:
    count = (df["trip_duration_minutes"] > threshold).sum()
    pct = count / len(df) * 100

    print(
        f"> {threshold:4} min: "
        f"{count:>8,} rides ({pct:.3f}%)"
    )
```

```python
df.groupby("rider_type")["trip_duration_minutes"].agg(
    ["count", "mean", "median", "max"]
)
```

```python
tableau_df = df[
    df["trip_duration_minutes"] <= 1440
].copy()

tableau_df.shape
```

```python
tableau_df.groupby("rider_type")["trip_duration_minutes"].agg(
    ["count", "mean", "median", "max"]
)
```

```python
tableau_columns = [
    "trip_id",
    "start_time",
    "from_station_id",
    "from_station_name",
    "to_station_id",
    "to_station_name",
    "rider_type",
    "trip_duration_minutes",
    "day_of_week",
    "month",
    "hour",
    "season"
]

tableau_df = tableau_df[tableau_columns].copy()

tableau_df.info()
```

```python
tableau_df["hour_label"] = pd.to_datetime(
    tableau_df["hour"],
    format="%H"
).dt.strftime("%-I %p")
```

```python
tableau_df[["hour", "hour_label"]].drop_duplicates().sort_values("hour")
```

```python
tableau_df.info()
```

```python
output_path = file_path / "cyclistic_2019_tableau.csv"

tableau_df.to_csv(output_path, index=False)
```

```python
file_size_mb = output_path.stat().st_size / (1024 ** 2)

print(f"Tableau dataset: {file_size_mb:.1f} MB")
```
